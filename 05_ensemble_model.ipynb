```python
from google.colab import drive
drive.mount('/content/drive')

import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns

from sklearn.metrics import (
    classification_report,
    confusion_matrix,
    accuracy_score,
    precision_score,
    recall_score,
    f1_score
)

from sklearn.pipeline import Pipeline
from sklearn.preprocessing import StandardScaler
from sklearn.linear_model import LogisticRegression
from sklearn.ensemble import RandomForestClassifier, VotingClassifier
from xgboost import XGBClassifier


PROJECT_ROOT = "/content/drive/MyDrive/student-dropout-prediction"

X_train = pd.read_csv(f"{PROJECT_ROOT}/data/processed/X_train.csv")
y_train = pd.read_csv(f"{PROJECT_ROOT}/data/processed/y_train.csv").values.ravel()

X_test = pd.read_csv(f"{PROJECT_ROOT}/data/processed/X_test.csv")
y_test = pd.read_csv(f"{PROJECT_ROOT}/data/processed/y_test.csv").values.ravel()
```

```python
# Logistic Regression pipeline (scaled)
lr_pipeline = Pipeline([
    ('scaler', StandardScaler()),
    ('lr', LogisticRegression(
        max_iter=5000,
        class_weight='balanced',
        random_state=42
    ))
])

# Random Forest (no scaling)
rf_pipeline = RandomForestClassifier(
    n_estimators=300,
    max_depth=10,
    min_samples_split=5,
    min_samples_leaf=2,
    class_weight='balanced',
    random_state=42,
    n_jobs=-1
)

# XGBoost (no scaling)
xgb_pipeline = XGBClassifier(
    n_estimators=200,
    max_depth=3,
    learning_rate=0.05,
    subsample=0.7,
    colsample_bytree=0.7,
    min_child_weight=10,
    gamma=0.5,
    scale_pos_weight=(y_train == 0).sum() / (y_train == 1).sum(),
    random_state=42,
    eval_metric='logloss'
)
```

```python
# VOTING ENSEMBLE - use Soft Voting
voting_clf = VotingClassifier(
    estimators=[
        ('lr', lr_pipeline),
        ('rf', rf_pipeline),
        ('xgb', xgb_pipeline)
    ],
    voting='soft'
)

voting_clf.fit(X_train, y_train)
```

```python
# Evaluate Ensemble
y_pred_ensemble = voting_clf.predict(X_test)

print(classification_report(y_test, y_pred_ensemble))
```

```python
# Confusion Matrix
cm_ensemble = confusion_matrix(y_test, y_pred_ensemble)

sns.heatmap(cm_ensemble, annot=True, fmt='d', cmap='Blues')
plt.xlabel("Predicted")
plt.ylabel("Actual")
plt.title("Ensemble Model – Confusion Matrix")
plt.show()
```

```python
tn, fp, fn, tp = cm_ensemble.ravel()

print(f"""
Ensemble Model Confusion Matrix:
--------------------------------
            Predicted
            0     1
Actual 0   {tn}    {fp}
Actual 1   {fn}    {tp}
""")
```

```python
from sklearn.metrics import roc_auc_score, roc_curve

y_prob_ensemble = voting_clf.predict_proba(X_test)[:, 1]
roc_auc_ensemble = roc_auc_score(y_test, y_prob_ensemble)

print("Ensemble ROC-AUC:", roc_auc_ensemble)

# ROC Curve
fpr, tpr, _ = roc_curve(y_test, y_prob_ensemble)

plt.figure()
plt.plot(fpr, tpr, label=f"ROC-AUC = {roc_auc_ensemble:.3f}")
plt.plot([0, 1], [0, 1], linestyle="--")
plt.xlabel("False Positive Rate")
plt.ylabel("True Positive Rate")
plt.title("Ensemble – ROC Curve")
plt.legend()
plt.show()
```

```python
# Save Ensemble Metrics
ensemble_results = pd.DataFrame({
    "Model": ["Ensemble (Voting)"],
    "Accuracy": [accuracy_score(y_test, y_pred_ensemble)],
    "Precision": [precision_score(y_test, y_pred_ensemble)],
    "Recall": [recall_score(y_test, y_pred_ensemble)],
    "F1-score": [f1_score(y_test, y_pred_ensemble)]
})

ensemble_results.to_csv(
    f"{PROJECT_ROOT}/results/metrics/ensemble_results.csv",
    index=False
)

ensemble_results
```

```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

# Combine results
results = pd.concat([lr, dt, rf, xgb, ens], ignore_index=True)

# Melt for grouped plotting
results_melted = results.melt(
    id_vars="Model",
    value_vars=["Accuracy", "Precision", "Recall", "F1-score"],
    var_name="Metric",
    value_name="Score"
)

# Find best model per metric
best_idx = results_melted.groupby("Metric")["Score"].idxmax()
results_melted["Best"] = False
results_melted.loc[best_idx, "Best"] = True

# Plot
plt.figure(figsize=(11, 6))

# Custom color palette (highlight best bars)
palette = []
for _, row in results_melted.iterrows():
    if row["Best"]:
        palette.append("#ff7f0e")  # highlight color (orange)
    else:
        palette.append("#1f77b4")  # normal color (blue)

ax = sns.barplot(
    data=results_melted,
    x="Model",
    y="Score",
    hue="Metric",
    dodge=True
)

# Add value labels on bars
for p in ax.patches:
    height = p.get_height()
    if not pd.isna(height):
        ax.annotate(
            f"{height*100:.1f}%",
            (p.get_x() + p.get_width() / 2, height),
            ha='center',
            va='bottom',
            fontsize=8,
            rotation=90,
            xytext=(0, 3),
            textcoords='offset points'
        )

plt.title("Model Evaluation Comparison (Highlighted Best Scores)")
plt.ylabel("Score (%)")
plt.xlabel("Model")
plt.ylim(0.7, 0.9)
plt.xticks(rotation=20)
plt.legend(title="Metric")
plt.tight_layout()
plt.show()
```

```python
# Accuracy comparison
plt.figure(figsize=(9,5))
sns.barplot(
    x="Model",
    y="Accuracy",
    data=results
)

plt.title("Accuracy Comparison of Machine Learning Models")
plt.ylabel("Accuracy")
plt.xlabel("Model")
plt.ylim(0.75, 0.9)
plt.xticks(rotation=20)
plt.tight_layout()
plt.show()
```
